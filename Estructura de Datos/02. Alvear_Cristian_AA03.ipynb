EJERCICIO 2: Sala de Espera (Lista Doble Enlazada)
Escenario: Gestión de pacientes que están esperando ser llamados. Se requiere
movimiento bidireccional.
• Clase TurnoEspera:
Atributos: numero_turno (ej. 101), nombre_paciente, hora_llegada, motivo_consulta.
• La Estructura a utilizar: Lista Doble con punteros sig y ant.
• Operaciones Obligatorias:
• llegada_paciente: Insertar al final.
• navegacion: Métodos para ver quién es el "Siguiente" y el "Anterior" a partir de un
paciente dado.
• Parte Personalizada: Crear una función de Prioridad de Emergencia. El programa
debe permitir mover un turno específico (por ID) directamente a la cabeza (inicio) de
la lista sin borrar el resto de los elementos.

```python
class TurnoEspera:
    def __init__(self, numero_turno, nombre_paciente, hora_llegada, motivo_consulta):
        self.numero_turno = numero_turno
        self.nombre_paciente = nombre_paciente
        self.hora_llegada = hora_llegada
        self.motivo_consulta = motivo_consulta
        
        # Punteros para el movimiento bidireccional
        self.sig = None  # Apunta al paciente que llegó después
        self.ant = None  # Apunta al paciente que llegó antes
```

```python
class SalaEspera:
    def __init__(self):
        self.cabeza = None # El primer paciente en ser atendido
        self.cola = None   # El último paciente en llegar

    # 1. Llegada de paciente (Insertar al final)
    def llegada_paciente(self, numero_turno, nombre_paciente, hora_llegada, motivo_consulta):
        nuevo_turno = TurnoEspera(numero_turno, nombre_paciente, hora_llegada, motivo_consulta)
        
        if self.cabeza is None: # Si la sala está vacía
            self.cabeza = nuevo_turno
            self.cola = nuevo_turno
        else: # Si ya hay gente, se pone al final de la cola
            self.cola.sig = nuevo_turno
            nuevo_turno.ant = self.cola
            self.cola = nuevo_turno

    # Función auxiliar sencilla para buscar un paciente por su turno
    def _buscar_nodo(self, numero_turno):
        actual = self.cabeza
        while actual:
            if actual.numero_turno == numero_turno:
                return actual
            actual = actual.sig
        return None

    # 2. Navegación: Ver Siguiente y Anterior
    def ver_vecinos(self, numero_turno):
        nodo = self._buscar_nodo(numero_turno)
        if nodo:
            anterior = nodo.ant.nombre_paciente if nodo.ant else "Nadie (es el primero)"
            siguiente = nodo.sig.nombre_paciente if nodo.sig else "Nadie (es el último)"
            print(f"[{nodo.nombre_paciente}] -> Anterior: {anterior} | Siguiente: {siguiente}")
        else:
            print("Turno no encontrado.")

    # 3. Prioridad de Emergencia (Mover a la cabeza)
    def prioridad_emergencia(self, numero_turno):
        nodo = self._buscar_nodo(numero_turno)
        
        # Si no existe o ya es el primero, no hacemos nada
        if not nodo or nodo == self.cabeza:
            return 

        # PASO A: Desconectar al paciente de su posición actual
        nodo.ant.sig = nodo.sig # El de atrás ahora apunta al de adelante
        
        if nodo.sig:
            nodo.sig.ant = nodo.ant # El de adelante ahora apunta al de atrás
        else:
            self.cola = nodo.ant # Si era el último de la fila, actualizamos la cola

        # PASO B: Mover el paciente al inicio de la fila
        nodo.sig = self.cabeza
        nodo.ant = None
        self.cabeza.ant = nodo
        self.cabeza = nodo

    # Función para imprimir y visualizar la sala
    def mostrar_sala(self):
        actual = self.cabeza
        print("\n--- SALA DE ESPERA ACTUAL ---")
        while actual:
            print(f"Turno {actual.numero_turno}: {actual.nombre_paciente} ({actual.motivo_consulta})")
            actual = actual.sig
        print("-----------------------------\n")
```

```python
# 1. Creamos la sala
mi_sala = SalaEspera()

# 2. Registramos llegadas (Insertando al final)
mi_sala.llegada_paciente(101, "Carlos", "08:00", "Control")
mi_sala.llegada_paciente(102, "Ana", "08:15", "Dolor de cabeza")
mi_sala.llegada_paciente(103, "Luis", "08:30", "Fractura de brazo")
mi_sala.llegada_paciente(104, "Marta", "08:45", "Examen de sangre")

# Mostramos cómo está la sala inicialmente
mi_sala.mostrar_sala()

# 3. Probamos la navegación
print("Navegación para Ana (Turno 102):")
mi_sala.ver_vecinos(102)

# 4. Probamos la Prioridad de Emergencia (Luis tiene una fractura)
print("\n¡EMERGENCIA! El turno 103 (Luis) pasa al frente.")
mi_sala.prioridad_emergencia(103)

# Mostramos la sala después de la emergencia
mi_sala.mostrar_sala()

# Verificamos cómo cambió la navegación de Luis y Ana
print("Navegación actualizada tras la emergencia:")
mi_sala.ver_vecinos(103) # Luis (ahora es el primero)
mi_sala.ver_vecinos(102) # Ana (su siguiente ahora es Marta)
```
