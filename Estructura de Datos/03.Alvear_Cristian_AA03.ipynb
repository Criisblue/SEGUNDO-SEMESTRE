EJERCICIO 3: Sistema de Monitoreo de Signos Vitales (Lista Doble Circular)
Escenario: En la Unidad de Cuidados Intensivos (UCI), un monitor central muestra
cíclicamente los signos vitales de los pacientes asignados a un enfermero. El monitor pasa
del Paciente A al B, luego al C, y al terminar con el último, vuelve automáticamente al
primero para reiniciar la vigilancia.
• Clase MonitoreoPaciente:
o id_cama: Entero (identificador de la cama).
o nombre_paciente: String.
o frecuencia_cardiaca: Entero (latidos por minuto).
o estado_alerta: String (ej. "Estable", "Crítico", "En Observación").
• La Estructura: Lista Doble Circular. No hay un final; el último paciente conectado
apunta al primero, permitiendo un monitoreo infinito en ambos sentidos
(siguiente/anterior).
• Operaciones Obligatorias:
o conectar_paciente: Insertar un nuevo paciente al ciclo de monitoreo.
o desconectar_paciente: Eliminar a un paciente (por id_cama) cuando es dado
de alta, asegurando que el ciclo se cierre correctamente entre el anterior y el
siguiente.
• Parte Personalizada: en esta parte se debe proponer una funcionalidad origina y
creativa adicional que aporte al sistema de la clínica. 

```python
class MonitoreoPaciente:
    def __init__(self, id_cama, nombre_paciente, frecuencia_cardiaca, estado_alerta):
        self.id_cama = id_cama
        self.nombre_paciente = nombre_paciente
        self.frecuencia_cardiaca = frecuencia_cardiaca
        self.estado_alerta = estado_alerta
        
        # Punteros para el ciclo infinito en ambas direcciones
        self.sig = None
        self.ant = None
```

```python
class MonitorUCI:
    def __init__(self):
        self.cabeza = None # Punto de partida del monitor

    def conectar_paciente(self, id_cama, nombre_paciente, frecuencia_cardiaca, estado_alerta):
        nuevo_paciente = MonitoreoPaciente(id_cama, nombre_paciente, frecuencia_cardiaca, estado_alerta)
        
        # Si el monitor está vacío, el paciente se apunta a sí mismo
        if self.cabeza is None:
            self.cabeza = nuevo_paciente
            self.cabeza.sig = self.cabeza
            self.cabeza.ant = self.cabeza
        else:
            # Encontrar el último paciente conectado (que está justo antes de la cabeza)
            ultimo = self.cabeza.ant
            
            # Conectar el nuevo paciente al final
            ultimo.sig = nuevo_paciente
            nuevo_paciente.ant = ultimo
            
            # Cerrar el círculo: conectar el nuevo con el primero
            nuevo_paciente.sig = self.cabeza
            self.cabeza.ant = nuevo_paciente
```

```python
def desconectar_paciente(self, id_cama):
        if self.cabeza is None:
            print("El monitor está vacío.")
            return False

        actual = self.cabeza
        while True:
            if actual.id_cama == id_cama:
                # Caso A: Es el único paciente en la UCI
                if actual.sig == self.cabeza and actual.ant == self.cabeza:
                    self.cabeza = None
                else:
                    # Caso B: Hay más pacientes. Se puentean los vecinos.
                    actual.ant.sig = actual.sig
                    actual.sig.ant = actual.ant
                    
                    # Si el paciente a borrar era el que estaba en pantalla (cabeza), 
                    # movemos la pantalla al siguiente
                    if actual == self.cabeza:
                        self.cabeza = actual.sig
                
                print(f"Paciente de la cama {id_cama} desconectado con éxito.")
                return True
            
            actual = actual.sig
            # Si dimos la vuelta completa y volvimos al inicio
            if actual == self.cabeza:
                break
                
        print(f"No se encontró la cama {id_cama}.")
        return False
```

```python
def escanear_alertas_criticas(self):
        if self.cabeza is None:
            return
            
        print("\n--- INICIANDO ESCANEO RÁPIDO DE TRIAGE ---")
        actual = self.cabeza
        alertas = 0
        
        # Recorremos el ciclo exactamente una vez
        while True:
            if actual.estado_alerta.upper() == "CRÍTICO":
                print(f"¡ALERTA CÓDIGO AZUL! Cama {actual.id_cama} - {actual.nombre_paciente} | FC: {actual.frecuencia_cardiaca} lpm")
                alertas += 1
                
            actual = actual.sig
            if actual == self.cabeza: # Terminamos la vuelta
                break
                
        if alertas == 0:
            print("Escaneo completado: Todos los pacientes están estables o en observación.")
        print("------------------------------------------\n")

    def mostrar_monitor_actual(self):
        if self.cabeza:
            print(f"Monitor mostrando -> Cama {self.cabeza.id_cama}: {self.cabeza.nombre_paciente} [{self.cabeza.estado_alerta}]")
```
