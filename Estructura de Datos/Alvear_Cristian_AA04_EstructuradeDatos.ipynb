```python
class NodoPedido:
    def __init__(self, id_pedido:int, tipo:str, descripcion:str):
        self.id_pedido = id_pedido
        self.tipo = tipo # Únicamente: "ESTÁNDAR" o "RECLAMO"
        self.descripcion = descripcion # Ej: "1 Hamburguesa con papas"
        self.siguiente = None # Puntero al siguiente Nodo en memoria
class GestorPedidos:
    def __init__(self):
        # Punteros para la Cola de Pedidos Estándar (FIFO)
        self.cola_frente = None
        self.cola_fin = None
        # Puntero para la Pila de Reclamos Urgentes (LIFO)
        self.pila_tope = None

# =====Mí código ==========
    def registrar_pedido(self, id_pedido:int, tipo:str,descripcion:str):
        nuevo_pedido=NodoPedido(id_pedido,tipo,descripcion)
## =====Caso si es ESTANDAR =====
        if tipo is "ESTANDAR":
            if self.cola_fin is None:
                self.cola_frente=nuevo_pedido
                self.cola_fin=nuevo_pedido
            else: #Cola con elementos
                self.cola_fin.siguiente=nuevo_pedido
                self.cola_fin=nuevo_pedido
## =====Caso si es RECLAMO =====
        else:
            if self.pila_tope is None:
                self.pila_tope=nuevo_pedido
            else: #Pilas con elementos
                nuevo_pedido.siguiente=self.pila_tope
                self.pila_tope=nuevo_pedido

### ========= AHORA SÍ LAS FUNCIONES ========
# LÓGICA QUE YO ENTENDÍ: Mientras haya reclamos, siempre se atenderá primero eso.
#                        Solo cuando no hay reclamos despacha ESTANDARES.
#                        Si no hay nada, se indicará que no hay pedidos.
    def esta_vacia(self):
        return self.cola_fin is None and self.pila_tope is None

    def despachar_siguiente(self):
        if self.esta_vacia():
            print("No hay pedidos para despachar.")
            return None
        if self.pila_tope is not None: 
            pedido_despachado=self.pila_tope
            self.pila_tope=self.pila_tope.siguiente
            return 
        else: # Despachar de la cola estándar
            pedido_despachado=self.cola_frente
            self.cola_frente=self.cola_frente.siguiente
            if self.cola_frente is None: # Si la cola queda vacía, actualizar el puntero de fin
                self.cola_fin = None
                return
    def activar_contingencia_masiva(self):
        while self.cola_fin is not None:
            nodo_paso=self.cola_fin
            self.cola_fin=self.cola_fin.siguiente
            nodo_paso=self.pila_tope
            self.pila_tope=self.pila_tope.siguiente
    def mostrar_pedidos(self):
        print("Pedidos en Cola Estándar:")
        nodo_actual = self.cola_frente
        while nodo_actual is not None:
            print(f"ID: {nodo_actual.id_pedido}, Tipo: {nodo_actual.tipo}, Descripción: {nodo_actual.descripcion}")
            nodo_actual = nodo_actual.siguiente
        print("\nPedidos en Pila de Reclamos Urgentes:")
        nodo_actual = self.pila_tope
        while nodo_actual is not None:
            print(f"ID: {nodo_actual.id_pedido}, Tipo: {nodo_actual.tipo}, Descripción: {nodo_actual.descripcion}")
            nodo_actual = nodo_actual.siguiente   
```

```python
class NodoPedido:
    def __init__(self, id_pedido: int, tipo: str, descripcion: str):
        self.id_pedido = id_pedido
        self.tipo = tipo
        self.descripcion = descripcion
        self.siguiente = None


class GestorPedidos:
    def __init__(self):
        self.cola_frente = None
        self.cola_fin = None
        self.pila_tope = None

    def registrar_pedido(self, id_pedido: int, tipo: str, descripcion: str):
        nuevo_pedido = NodoPedido(id_pedido, tipo, descripcion)
        
        if tipo == "ESTANDAR":
            if self.cola_fin is None:
                self.cola_frente = nuevo_pedido
                self.cola_fin = nuevo_pedido
            else:
                self.cola_fin.siguiente = nuevo_pedido
                self.cola_fin = nuevo_pedido
        else:
            nuevo_pedido.siguiente = self.pila_tope
            self.pila_tope = nuevo_pedido

    def esta_vacia(self):
        return self.cola_frente is None and self.pila_tope is None

    def despachar_siguiente(self):
        if self.esta_vacia():
            print("No hay pedidos para despachar.")
            return None
        
        if self.pila_tope is not None:
            pedido_despachado = self.pila_tope
            self.pila_tope = self.pila_tope.siguiente
            print(f"Despachado RECLAMO - ID: {pedido_despachado.id_pedido}")
            return pedido_despachado
        else:
            pedido_despachado = self.cola_frente
            self.cola_frente = self.cola_frente.siguiente
            if self.cola_frente is None:
                self.cola_fin = None
            print(f"Despachado ESTANDAR - ID: {pedido_despachado.id_pedido}")
            return pedido_despachado

    def activar_contingencia_masiva(self):
        while self.cola_frente is not None:
            nodo_trasladado = self.cola_frente
            self.cola_frente = self.cola_frente.siguiente
            nodo_trasladado.siguiente = self.pila_tope
            self.pila_tope = nodo_trasladado
        
        self.cola_fin = None
        print("Contingencia masiva activada. Pedidos estandar trasladados a reclamos.")

    def mostrar_pedidos(self):
        print("\n" + "-" * 50)
        print("COLA DE PEDIDOS ESTANDAR")
        print("-" * 50)
        
        nodo_actual = self.cola_frente
        if nodo_actual is None:
            print("Vacia")
        else:
            contador = 1
            while nodo_actual is not None:
                print(f"{contador}. ID: {nodo_actual.id_pedido} - {nodo_actual.descripcion}")
                nodo_actual = nodo_actual.siguiente
                contador += 1

        print("\n" + "-" * 50)
        print("PILA DE RECLAMOS URGENTES")
        print("-" * 50)
        
        nodo_actual = self.pila_tope
        if nodo_actual is None:
            print("Vacia")
        else:
            contador = 1
            while nodo_actual is not None:
                print(f"{contador}. ID: {nodo_actual.id_pedido} - {nodo_actual.descripcion}")
                nodo_actual = nodo_actual.siguiente
                contador += 1
        
        print("-" * 50 + "\n")


def main():
    gestor = GestorPedidos()
    
    while True:
        print("\n" + "=" * 50)
        print("GESTOR DE PEDIDOS")
        print("=" * 50)
        print("1. Registrar pedido ESTANDAR")
        print("2. Registrar pedido RECLAMO")
        print("3. Despachar siguiente pedido")
        print("4. Activar contingencia masiva")
        print("5. Ver todos los pedidos")
        print("6. Salir")
        print("=" * 50)
        
        opcion = input("Selecciona una opcion (1-6): ").strip()
        
        if opcion == "1":
            try:
                id_pedido = int(input("ID del pedido: "))
                descripcion = input("Descripcion: ")
                gestor.registrar_pedido(id_pedido, "ESTANDAR", descripcion)
                print("Pedido estandar registrado.")
            except ValueError:
                print("Error: ID debe ser un numero.")
        
        elif opcion == "2":
            try:
                id_pedido = int(input("ID del pedido: "))
                descripcion = input("Descripcion: ")
                gestor.registrar_pedido(id_pedido, "RECLAMO", descripcion)
                print("Reclamo registrado.")
            except ValueError:
                print("Error: ID debe ser un numero.")
        
        elif opcion == "3":
            gestor.despachar_siguiente()
        
        elif opcion == "4":
            if gestor.cola_frente is None:
                print("No hay pedidos estandar para trasladar.")
            else:
                gestor.activar_contingencia_masiva()
        
        elif opcion == "5":
            gestor.mostrar_pedidos()
        
        elif opcion == "6":
            print("Sistema cerrado.")
            break
        
        else:
            print("Opcion no valida.")


if __name__ == "__main__":
    main()
```
