## Ejercicio 1: Historial de Atenciones (Lista Enlazada Simple)

En este ejercicio se registra de forma cronológica a los pacientes que ya terminaron su consulta. Para esto se usa una **lista enlazada simple** donde cada nuevo registro se inserta al final.

No se usan listas nativas de Python `[]`, todo se maneja con nodos y punteros (`sig`).

### Clase Nodo
Cada nodo guarda una atención médica y un puntero al siguiente nodo.

```python
class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.sig = None
```

### Clase AtencionMedica
Aquí se guardan los datos de cada atención: id, nombre del paciente, fecha y monto.

```python
class AtencionMedica:
    def __init__(self, id_registro, nombre_paciente, fecha, monto_consulta):
        self.id_registro = id_registro
        self.nombre_paciente = nombre_paciente
        self.fecha = fecha
        self.monto_consulta = monto_consulta

    def mostrar(self):
        print(f"ID: {self.id_registro} | Paciente: {self.nombre_paciente} | Fecha: {self.fecha} | Monto: ${self.monto_consulta}")
```

### Clase Historial (Lista Enlazada Simple)
Esta es la estructura principal. Tiene los métodos para:
- Insertar al final
- Mostrar todo el historial
- Filtrar por monto mínimo

```python
class Historial:
    def __init__(self):
        self.cabeza = None

    # Insertar siempre al final
    def insertar(self, atencion):
        nuevo = Nodo(atencion)
        if self.cabeza is None:
            self.cabeza = nuevo
        else:
            actual = self.cabeza
            while actual.sig is not None:
                actual = actual.sig
            actual.sig = nuevo
        print(f"Registro {atencion.id_registro} agregado al historial.")

    # Recorrer y mostrar todo
    def mostrar_historial(self):
        if self.cabeza is None:
            print("El historial está vacío.")
            return
        print("\n--- HISTORIAL DE ATENCIONES ---")
        actual = self.cabeza
        while actual is not None:
            actual.dato.mostrar()
            actual = actual.sig

    # Filtro personalizado: monto mayor al valor ingresado
    def filtrar_por_monto(self, valor_minimo):
        if self.cabeza is None:
            print("El historial está vacío.")
            return
        print(f"\n--- Consultas con monto mayor a ${valor_minimo} ---")
        actual = self.cabeza
        contador = 0
        while actual is not None:
            if actual.dato.monto_consulta > valor_minimo:
                actual.dato.mostrar()
                contador += 1
            actual = actual.sig
        if contador == 0:
            print("No hay consultas que cumplan ese criterio.")
```

### Pruebas
Creamos el historial y agregamos algunas atenciones de ejemplo.

```python
historial = Historial()

a1 = AtencionMedica(1, "Juan Perez", "10/05/2025", 25.00)
a2 = AtencionMedica(2, "Maria Lopez", "11/05/2025", 50.00)
a3 = AtencionMedica(3, "Carlos Ruiz", "12/05/2025", 80.00)
a4 = AtencionMedica(4, "Ana Torres", "13/05/2025", 35.00)
a5 = AtencionMedica(5, "Pedro Gomez", "14/05/2025", 120.00)

historial.insertar(a1)
historial.insertar(a2)
historial.insertar(a3)
historial.insertar(a4)
historial.insertar(a5)
```

Ahora mostramos todo el historial recorriendo la lista nodo por nodo.

```python
historial.mostrar_historial()
```

### Filtro por monto
Le pedimos al usuario un valor mínimo por teclado y se muestran solo las consultas con monto mayor.

```python
valor = float(input("Ingrese el monto mínimo para filtrar: "))
historial.filtrar_por_monto(valor)
```

### Conclusión
Con la lista enlazada simple se pudo registrar las atenciones en orden cronológico, mostrar el historial completo y filtrar solo las consultas que superan cierto monto, todo usando nodos y punteros sin listas nativas.
