## Caso de estudio: consumo energético de dispositivos tecnológicos

### 1. Creación del arreglo NumPy

```python
import numpy as np

consumo = np.array([
    [12.5, 13.0, 12.8, 14.1, 13.7, 6.2, 4.8],   # Computadoras
    [6.5,  6.8,  7.0,  6.9,  7.2,  3.1, 2.5],   # Laptops
    [18.0, 18.2, 18.1, 18.3, 18.4, 18.0, 17.9], # Servidor local
    [4.2,  4.1,  4.3,  4.4,  4.2,  4.0,  3.9],  # Red
    [9.0,  9.5,  9.2,  9.8,  10.0, 3.5,  2.0]   # Iluminación
])

print("Matriz completa:")
print(consumo)
print("\nNúmero de dimensiones:", consumo.ndim)
print("Forma del arreglo:", consumo.shape)
print("Número total de elementos:", consumo.size)
print("Tipo de dato:", consumo.dtype)
```

### 2. Análisis general del consumo

```python
total_semanal = np.sum(consumo)
promedio_general = np.mean(consumo)
maximo = np.max(consumo)
minimo = np.min(consumo)

print("Consumo total semanal (todos los dispositivos):", total_semanal, "kWh")
print("Consumo promedio general:", round(promedio_general, 2), "kWh")
print("Consumo máximo registrado:", maximo, "kWh")
print("Consumo mínimo registrado:", minimo, "kWh")
```

**Interpretación:** El consumo energético varía considerablemente entre dispositivos. El servidor local es el que más contribuye al total semanal, manteniéndose elevado todos los días. Los días de fin de semana muestran una reducción notable en la mayoría de dispositivos, lo que indica que el uso de la oficina disminuye. En general, el promedio se ve influenciado principalmente por el servidor local.

### 3. Análisis por dispositivo usando axis

```python
dispositivos = ["Computadoras", "Laptops", "Servidor local", "Red", "Iluminación"]

total_por_dispositivo = np.sum(consumo, axis=1)

for i, nombre in enumerate(dispositivos):
    print(f"{nombre}: {total_por_dispositivo[i]} kWh")

print("\nDispositivo con mayor consumo:", dispositivos[np.argmax(total_por_dispositivo)])
print("Dispositivo con menor consumo:", dispositivos[np.argmin(total_por_dispositivo)])

std_por_dispositivo = np.std(consumo, axis=1)
print("\nDesviación estándar por dispositivo:")
for i, nombre in enumerate(dispositivos):
    print(f"  {nombre}: {round(std_por_dispositivo[i], 4)}")
print("Dispositivo más constante:", dispositivos[np.argmin(std_por_dispositivo)])
```

**Respuestas:**

1. El dispositivo que consumió más energía durante la semana fue el **Servidor local**.
2. El dispositivo que consumió menos energía fue la **Red**.
3. El dispositivo con consumo más constante es el **Servidor local**, ya que su desviación estándar es la más baja en términos relativos a su consumo.
4. El servidor local mantiene un consumo prácticamente igual todos los días, incluidos fines de semana. Esto indica que opera de forma continua sin importar si la oficina está en uso o no, lo cual representa una oportunidad de ahorro si se puede programar su apagado o reducción de carga fuera del horario laboral.

### 4. Análisis por día usando axis

```python
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]

total_por_dia = np.sum(consumo, axis=0)

for i, dia in enumerate(dias):
    print(f"{dia}: {round(total_por_dia[i], 2)} kWh")

print("\nDía con mayor consumo:", dias[np.argmax(total_por_dia)])
print("Día con menor consumo:", dias[np.argmin(total_por_dia)])
```

**Respuestas:**

1. El día con mayor consumo energético fue el **Jueves**.
2. El día con menor consumo energético fue el **Domingo**.
3. Los días laborables (lunes a viernes) tienen un consumo significativamente mayor que el fin de semana. El sábado y el domingo presentan valores mucho menores en casi todos los dispositivos, excepto en el servidor local.

### 5. Transformación del arreglo

```python
consumo_1d = consumo.flatten()

print("Arreglo unidimensional:")
print(consumo_1d)
print("\nNúmero de datos:", consumo_1d.size)

promedio_1d = np.mean(consumo_1d)
print("Promedio desde arreglo 1D:", round(promedio_1d, 2), "kWh")
print("Promedio desde matriz original:", round(promedio_general, 2), "kWh")
print("¿Son iguales?", round(promedio_1d, 10) == round(promedio_general, 10))
```

**Respuestas:**

1. El arreglo transformado contiene los 35 valores del consumo en una sola fila.
2. El arreglo contiene **35 datos** (5 dispositivos × 7 días).
3. El promedio general calculado desde el arreglo 1D coincide con el obtenido desde la matriz.
4. Ambos resultados son iguales porque la operación `mean` calcula el promedio sobre todos los elementos sin importar la forma del arreglo. Al transformar la matriz a una dimensión no se pierde ni se agrega ningún dato, solo cambia la estructura.
5. Esto muestra que la forma del arreglo no afecta los cálculos globales; NumPy opera sobre los datos independientemente de cómo estén organizados.

### 7. Reflexión final

NumPy es una herramienta útil para analizar datos numéricos porque permite realizar operaciones matemáticas sobre grandes conjuntos de datos de forma rápida y con muy pocas líneas de código. A diferencia de las listas tradicionales de Python, los arreglos de NumPy están optimizados para cálculos numéricos, lo que los hace más eficientes tanto en velocidad como en uso de memoria. Además, NumPy permite aplicar funciones como `sum`, `mean`, `max` o `min` directamente sobre el arreglo sin necesidad de usar ciclos.

El parámetro `axis` es especialmente útil cuando los datos están organizados en filas y columnas, ya que permite calcular resultados por fila o por columna de forma directa. En este caso, se pudo obtener el consumo total por dispositivo (axis=1) y por día (axis=0) sin reorganizar los datos.

Este tipo de análisis puede aplicarse en situaciones como el seguimiento de ventas por sucursal y por semana, el monitoreo de temperatura en distintos puntos de una red de sensores, el análisis de calificaciones de estudiantes por materia y por periodo, o el control de inventario en una bodega. En general, cualquier problema donde los datos estén organizados en una tabla numérica puede beneficiarse del uso de NumPy.
