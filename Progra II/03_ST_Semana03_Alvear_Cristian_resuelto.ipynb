## Ejercicio 1. Sistema de empleados universitarios

Desarrolle un sistema orientado a objetos para representar personal universitario.

### Requisitos
- Cree una clase base llamada `Empleado` con:
  - `nombre`
  - `sueldo`
  - método `mostrar_datos()`
  - método `calcular_bono()`

- Cree dos clases hijas:
  - `Docente`
  - `Administrativo`

### Condiciones
- En `Docente`, agregue el atributo `horas_clase`.
- En `Administrativo`, agregue el atributo `departamento`.
- Sobrescriba el método `calcular_bono()` en cada clase hija.
- Use `super()` donde corresponda.
- Cree varios objetos y recórralos en una lista.

### Analice
- ¿Qué atributos y métodos deben estar en la clase padre?
- ¿Qué comportamiento cambia según el tipo de empleado?

```python
# Ejercicio 1: Sistema de empleados universitarios

class Empleado:
    """Clase base que representa a cualquier empleado universitario."""

    def __init__(self, nombre, sueldo):
        self.nombre = nombre
        self.sueldo = sueldo

    def mostrar_datos(self):
        print(f"Nombre: {self.nombre}")
        print(f"Sueldo base: ${self.sueldo:.2f}")

    def calcular_bono(self):
        # Bono general por defecto: 5% del sueldo
        return self.sueldo * 0.05


class Docente(Empleado):
    """Empleado docente: gana un adicional por las horas de clase dictadas."""

    def __init__(self, nombre, sueldo, horas_clase):
        super().__init__(nombre, sueldo)
        self.horas_clase = horas_clase

    def mostrar_datos(self):
        super().mostrar_datos()
        print(f"Horas de clase: {self.horas_clase}")

    def calcular_bono(self):
        # Reutilizamos el bono base y le sumamos un adicional por hora
        bono_base = super().calcular_bono()
        adicional = self.horas_clase * 10
        return bono_base + adicional


class Administrativo(Empleado):
    """Empleado administrativo asignado a un departamento."""

    def __init__(self, nombre, sueldo, departamento):
        super().__init__(nombre, sueldo)
        self.departamento = departamento

    def mostrar_datos(self):
        super().mostrar_datos()
        print(f"Departamento: {self.departamento}")

    def calcular_bono(self):
        # Bono fijo del 8% del sueldo
        return self.sueldo * 0.08


# Lista polimórfica de empleados
empleados = [
    Docente("Ana Martínez", 1200, 20),
    Docente("Luis Pérez", 1500, 25),
    Administrativo("Carla Ríos", 1000, "Talento Humano"),
    Administrativo("Mario Soto", 1100, "Contabilidad"),
]

for emp in empleados:
    emp.mostrar_datos()
    print(f"Bono: ${emp.calcular_bono():.2f}")
    print("-" * 40)

# Análisis:
# - En la clase padre van los atributos y métodos comunes a TODO empleado:
#   nombre, sueldo, mostrar_datos() y un calcular_bono() genérico.
# - Lo que cambia entre tipos es CÓMO se calcula el bono y qué datos extra
#   tiene cada uno (horas_clase para Docente, departamento para Administrativo).
```

## Ejercicio 2. Sistema de figuras con cálculo de área

Construya una jerarquía de clases para representar figuras geométricas.

### Requisitos
- Cree una clase base `Figura` con:
  - método `calcular_area()`
  - método `describir()`

- Cree las clases hijas:
  - `Rectangulo`
  - `Circulo`
  - `Triangulo`

### Condiciones
- Cada clase hija debe sobrescribir `calcular_area()`.
- Cada clase hija debe sobrescribir `describir()`.
- Recorra una lista de figuras y muestre:
  - el tipo de figura,
  - sus datos,
  - su área.

### Reto adicional
- Investigue cómo redondear resultados a 2 decimales.

```python
# Ejercicio 2: Sistema de figuras con cálculo de área
import math


class Figura:
    """Clase base abstracta para figuras geométricas."""

    def calcular_area(self):
        return 0

    def describir(self):
        return "Figura geométrica genérica"


class Rectangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return round(self.base * self.altura, 2)

    def describir(self):
        return f"Rectángulo de base {self.base} y altura {self.altura}"


class Circulo(Figura):
    def __init__(self, radio):
        self.radio = radio

    def calcular_area(self):
        # round(valor, 2) redondea a 2 decimales (reto adicional)
        return round(math.pi * self.radio ** 2, 2)

    def describir(self):
        return f"Círculo de radio {self.radio}"


class Triangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return round((self.base * self.altura) / 2, 2)

    def describir(self):
        return f"Triángulo de base {self.base} y altura {self.altura}"


# Lista de figuras y recorrido polimórfico
figuras = [
    Rectangulo(5, 3),
    Circulo(4),
    Triangulo(6, 8),
]

for f in figuras:
    print(f"Tipo de figura: {type(f).__name__}")
    print(f"Datos: {f.describir()}")
    print(f"Área: {f.calcular_area()}")
    print("-" * 40)
```

## Ejercicio 3. Gestión de cuentas bancarias

Desarrolle un sistema de cuentas bancarias usando herencia.

### Requisitos
- Cree una clase base `CuentaBancaria` con:
  - `titular`
  - `saldo`
  - método `depositar(monto)`
  - método `retirar(monto)`
  - método `mostrar_info()`

- Cree dos clases hijas:
  - `CuentaAhorros`
  - `CuentaCorriente`

### Condiciones
- En `CuentaAhorros`, no permita retirar más dinero del saldo disponible.
- En `CuentaCorriente`, permita sobregiro hasta un valor límite.
- Sobrescriba el método `retirar()` en ambas clases hijas.
- Valide que no se ingresen montos negativos.
- Muestre mensajes claros al usuario.

### Analice
- ¿Por qué este problema requiere sobrescritura?
- ¿Qué reglas cambian de una cuenta a otra?

```python
# Ejercicio 3: Gestión de cuentas bancarias

class CuentaBancaria:
    """Clase base con la lógica común de una cuenta."""

    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, monto):
        if monto <= 0:
            print("Error: el monto debe ser positivo.")
            return
        self.saldo += monto
        print(f"Depósito exitoso. Saldo actual: ${self.saldo:.2f}")

    def retirar(self, monto):
        if monto <= 0:
            print("Error: el monto debe ser positivo.")
            return
        self.saldo -= monto
        print(f"Retiro exitoso. Saldo actual: ${self.saldo:.2f}")

    def mostrar_info(self):
        print(f"Titular: {self.titular}")
        print(f"Saldo: ${self.saldo:.2f}")


class CuentaAhorros(CuentaBancaria):
    """No permite retirar más de lo que hay en saldo."""

    def retirar(self, monto):
        if monto <= 0:
            print("Error: el monto debe ser positivo.")
            return
        if monto > self.saldo:
            print(f"Error: saldo insuficiente. Disponible: ${self.saldo:.2f}")
            return
        super().retirar(monto)

    def mostrar_info(self):
        print("--- Cuenta de Ahorros ---")
        super().mostrar_info()


class CuentaCorriente(CuentaBancaria):
    """Permite sobregiro hasta un límite definido."""

    def __init__(self, titular, saldo=0, limite_sobregiro=500):
        super().__init__(titular, saldo)
        self.limite_sobregiro = limite_sobregiro

    def retirar(self, monto):
        if monto <= 0:
            print("Error: el monto debe ser positivo.")
            return
        if monto > self.saldo + self.limite_sobregiro:
            print(f"Error: excede el límite de sobregiro de ${self.limite_sobregiro:.2f}.")
            return
        super().retirar(monto)

    def mostrar_info(self):
        print("--- Cuenta Corriente ---")
        super().mostrar_info()
        print(f"Límite de sobregiro: ${self.limite_sobregiro:.2f}")


# Pruebas
print(">> Cuenta de Ahorros")
ahorro = CuentaAhorros("Cristian Alvear", 500)
ahorro.mostrar_info()
ahorro.depositar(200)
ahorro.retirar(800)    # debe rechazar (no alcanza)
ahorro.retirar(500)    # OK
ahorro.retirar(-50)    # rechazo por monto negativo

print()
print(">> Cuenta Corriente")
corriente = CuentaCorriente("María López", 300, 400)
corriente.mostrar_info()
corriente.retirar(500)   # usa parte del sobregiro -> queda en -200
corriente.retirar(300)   # excede el sobregiro
corriente.depositar(100)

# Análisis:
# - El problema requiere sobrescritura porque cada tipo de cuenta tiene
#   reglas DISTINTAS para el método retirar(): ahorros bloquea si no hay
#   saldo, corriente permite ir a negativo hasta cierto límite.
# - Lo que cambia: la validación previa al retiro y, en corriente, el
#   atributo extra limite_sobregiro.
```

## Ejercicio 4. Inventario de dispositivos tecnológicos

Modele un inventario de dispositivos usando herencia y polimorfismo.

### Requisitos
- Cree una clase base `Dispositivo` con:
  - `marca`
  - `modelo`
  - método `encender()`
  - método `mostrar_info()`

- Cree las clases hijas:
  - `Laptop`
  - `Telefono`
  - `Tablet`

### Condiciones
- Cada clase hija debe redefinir el método `encender()`.
- Agregue al menos un atributo exclusivo en cada clase hija.
- Cree una lista con varios dispositivos.
- Recorra la lista y ejecute métodos comunes para demostrar polimorfismo.

### Reto adicional
- Agregue un método `tipo_uso()` que cambie según el dispositivo.

```python
# Ejercicio 4: Inventario de dispositivos tecnológicos

class Dispositivo:
    """Clase base de cualquier dispositivo electrónico."""

    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def encender(self):
        print(f"{self.marca} {self.modelo} se está encendiendo...")

    def mostrar_info(self):
        print(f"Marca: {self.marca} | Modelo: {self.modelo}")

    def tipo_uso(self):
        return "Uso general"


class Laptop(Dispositivo):
    def __init__(self, marca, modelo, ram_gb):
        super().__init__(marca, modelo)
        self.ram_gb = ram_gb

    def encender(self):
        print(f"Laptop {self.marca} {self.modelo} cargando sistema operativo...")

    def mostrar_info(self):
        super().mostrar_info()
        print(f"RAM: {self.ram_gb} GB")

    def tipo_uso(self):
        return "Trabajo, programación y estudio"


class Telefono(Dispositivo):
    def __init__(self, marca, modelo, sistema_operativo):
        super().__init__(marca, modelo)
        self.sistema_operativo = sistema_operativo

    def encender(self):
        print(f"Teléfono {self.marca} {self.modelo} mostrando logo de {self.sistema_operativo}...")

    def mostrar_info(self):
        super().mostrar_info()
        print(f"SO: {self.sistema_operativo}")

    def tipo_uso(self):
        return "Comunicación y redes sociales"


class Tablet(Dispositivo):
    def __init__(self, marca, modelo, pulgadas):
        super().__init__(marca, modelo)
        self.pulgadas = pulgadas

    def encender(self):
        print(f"Tablet {self.marca} {self.modelo} iniciando pantalla táctil...")

    def mostrar_info(self):
        super().mostrar_info()
        print(f"Pantalla: {self.pulgadas} pulgadas")

    def tipo_uso(self):
        return "Lectura y multimedia"


# Lista polimórfica de dispositivos
dispositivos = [
    Laptop("HP", "Pavilion 15", 16),
    Telefono("Samsung", "Galaxy S24", "Android"),
    Tablet("Apple", "iPad Air", 10.9),
]

for d in dispositivos:
    d.mostrar_info()
    d.encender()
    print(f"Uso típico: {d.tipo_uso()}")
    print("-" * 40)
```

## Ejercicio 5. Plataforma académica con herencia multinivel

Modele una estructura académica con herencia multinivel.

### Requisitos
- Cree la clase `Persona`.
- Cree la clase `Estudiante` que herede de `Persona`.
- Cree la clase `EstudianteBecado` que herede de `Estudiante`.

### Condiciones
- `Persona` debe tener:
  - `nombre`
  - `cedula`

- `Estudiante` debe agregar:
  - `carrera`
  - `nivel`

- `EstudianteBecado` debe agregar:
  - `tipo_beca`
  - `porcentaje_descuento`

- Use `super()` correctamente en todos los niveles.
- Cree un método para mostrar toda la información.

### Analice
- ¿Qué atributos se heredan desde cada nivel?
- ¿Qué ventajas tiene esta jerarquía?
- ¿Qué desventajas tendría si crece demasiado?

```python
# Ejercicio 5: Plataforma académica con herencia multinivel

class Persona:
    def __init__(self, nombre, cedula):
        self.nombre = nombre
        self.cedula = cedula

    def mostrar_info(self):
        print(f"Nombre: {self.nombre}")
        print(f"Cédula: {self.cedula}")


class Estudiante(Persona):
    def __init__(self, nombre, cedula, carrera, nivel):
        super().__init__(nombre, cedula)
        self.carrera = carrera
        self.nivel = nivel

    def mostrar_info(self):
        super().mostrar_info()
        print(f"Carrera: {self.carrera}")
        print(f"Nivel: {self.nivel}")


class EstudianteBecado(Estudiante):
    def __init__(self, nombre, cedula, carrera, nivel, tipo_beca, porcentaje_descuento):
        super().__init__(nombre, cedula, carrera, nivel)
        self.tipo_beca = tipo_beca
        self.porcentaje_descuento = porcentaje_descuento

    def mostrar_info(self):
        super().mostrar_info()
        print(f"Tipo de beca: {self.tipo_beca}")
        print(f"Descuento: {self.porcentaje_descuento}%")


# Pruebas
estudiante1 = Estudiante("Pedro Vera", "0911223344", "Software", 4)
becado1 = EstudianteBecado(
    "Cristian Alvear", "0987654321", "Software", 3,
    "Académica", 50
)

print(">> Estudiante regular")
estudiante1.mostrar_info()
print()
print(">> Estudiante becado")
becado1.mostrar_info()

# Análisis:
# - Desde Persona se hereda nombre y cedula.
# - Desde Estudiante se heredan además carrera y nivel.
# - EstudianteBecado añade tipo_beca y porcentaje_descuento.
# Ventajas: máxima reutilización; cada clase añade solo lo nuevo.
# Desventajas si crece mucho: jerarquía rígida y frágil; un cambio en
# Persona afecta a TODOS los hijos. Para sistemas grandes suele ser
# preferible la composición o usar mixins.
```

## Ejercicio 6. Polimorfismo con reporte general

Desarrolle un sistema en el que distintos objetos respondan al mismo método.

### Requisitos
- Cree una clase base `UsuarioSistema`.
- Cree las clases hijas:
  - `Estudiante`
  - `Docente`
  - `Coordinador`

### Condiciones
- Todas las clases deben implementar el método `generar_reporte()`.
- Cada reporte debe mostrar información distinta según el rol.
- Cree una lista con varios usuarios.
- Recorra la lista y llame al método `generar_reporte()`.

### Reto adicional
- Ordene los objetos por nombre antes de imprimir los reportes.

```python
# Ejercicio 6: Polimorfismo con reporte general

class UsuarioSistema:
    """Clase base de cualquier usuario del sistema."""

    def __init__(self, nombre):
        self.nombre = nombre

    def generar_reporte(self):
        print(f"Reporte general del usuario {self.nombre}")


class Estudiante(UsuarioSistema):
    def __init__(self, nombre, promedio):
        super().__init__(nombre)
        self.promedio = promedio

    def generar_reporte(self):
        print(f"[ESTUDIANTE] {self.nombre} - Promedio: {self.promedio}")


class Docente(UsuarioSistema):
    def __init__(self, nombre, materias):
        super().__init__(nombre)
        self.materias = materias

    def generar_reporte(self):
        print(f"[DOCENTE] {self.nombre} - Materias dictadas: {self.materias}")


class Coordinador(UsuarioSistema):
    def __init__(self, nombre, area):
        super().__init__(nombre)
        self.area = area

    def generar_reporte(self):
        print(f"[COORDINADOR] {self.nombre} - Área a cargo: {self.area}")


# Lista de usuarios variados
usuarios = [
    Estudiante("Pedro Vera", 8.5),
    Docente("Ana Castro", 4),
    Coordinador("Luis Mora", "Tecnología"),
    Estudiante("Belén Ramos", 9.2),
]

# Reto adicional: ordenar por nombre antes de imprimir
usuarios_ordenados = sorted(usuarios, key=lambda u: u.nombre)

print(">> Reportes ordenados por nombre")
for u in usuarios_ordenados:
    u.generar_reporte()
```

## Ejercicio 7. Uso de `super()` en métodos no constructores

Construya un sistema donde `super()` se use también dentro de otros métodos.

### Requisitos
- Cree una clase base `Persona` con un método `presentarse()`.
- Cree una clase hija `Docente`.
- Cree una clase hija `Investigador`.

### Condiciones
- En las clases hijas, sobrescriba `presentarse()`.
- Use `super().presentarse()` para reutilizar parte del comportamiento.
- Luego agregue información específica en cada clase hija.

### Analice
- ¿Qué ventaja tiene extender un método en lugar de reemplazarlo completamente?

```python
# Ejercicio 7: super() en métodos no constructores

class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    def presentarse(self):
        print(f"Hola, mi nombre es {self.nombre}.")


class Docente(Persona):
    def __init__(self, nombre, materia):
        super().__init__(nombre)
        self.materia = materia

    def presentarse(self):
        # Reutilizamos el comportamiento del padre y AÑADIMOS información
        super().presentarse()
        print(f"Soy docente y dicto la materia de {self.materia}.")


class Investigador(Persona):
    def __init__(self, nombre, area_investigacion):
        super().__init__(nombre)
        self.area_investigacion = area_investigacion

    def presentarse(self):
        super().presentarse()
        print(f"Trabajo como investigador en el área de {self.area_investigacion}.")


# Pruebas
personas = [
    Docente("Ing. Rocío Mendoza", "Programación II"),
    Investigador("Dr. Carlos Vélez", "Inteligencia Artificial"),
]
for p in personas:
    p.presentarse()
    print("-" * 40)

# Análisis:
# Extender un método con super() en lugar de reemplazarlo:
# - Evita duplicar código (DRY).
# - Si el padre cambia, los hijos heredan automáticamente la mejora.
# - Permite especializar SOLO la parte que difiere de la lógica base.
```

## Ejercicio 8. Reto integrador

Desarrolle un mini sistema orientado a objetos relacionado con un contexto real.

### Opciones
Puede elegir uno de estos contextos:
- sistema de transporte,
- sistema hospitalario,
- tienda virtual,
- gestión académica,
- biblioteca.

### Requisitos mínimos
- una clase base,
- al menos tres clases hijas,
- uso de `super()`,
- sobrescritura de métodos,
- polimorfismo con lista de objetos,
- salida clara y ordenada.

### Producto esperado
El programa debe demostrar claramente:
- reutilización,
- especialización,
- comportamiento polimórfico.

```python
# Ejercicio 8: Reto integrador - Sistema de transporte
# Contexto elegido: una empresa de transporte que opera distintos vehículos
# y necesita calcular tarifas según el tipo de vehículo.


class Vehiculo:
    """Clase base de cualquier vehículo de la flota."""

    def __init__(self, placa, marca, capacidad):
        self.placa = placa
        self.marca = marca
        self.capacidad = capacidad

    def describir(self):
        print(f"Placa: {self.placa} | Marca: {self.marca} | Capacidad: {self.capacidad}")

    def calcular_tarifa(self, km):
        # Tarifa genérica por km
        return km * 0.50


class Bus(Vehiculo):
    def __init__(self, placa, marca, capacidad, ruta):
        super().__init__(placa, marca, capacidad)
        self.ruta = ruta

    def describir(self):
        print(">> BUS")
        super().describir()
        print(f"Ruta: {self.ruta}")

    def calcular_tarifa(self, km):
        # Tarifa social: $0.30 por km recorrido
        return 0.30 * km


class Taxi(Vehiculo):
    def __init__(self, placa, marca, capacidad, conductor):
        super().__init__(placa, marca, capacidad)
        self.conductor = conductor

    def describir(self):
        print(">> TAXI")
        super().describir()
        print(f"Conductor: {self.conductor}")

    def calcular_tarifa(self, km):
        # Banderazo de $1.50 + $0.80 por km
        return 1.50 + (km * 0.80)


class Camion(Vehiculo):
    def __init__(self, placa, marca, capacidad, carga_max_ton):
        super().__init__(placa, marca, capacidad)
        self.carga_max_ton = carga_max_ton

    def describir(self):
        print(">> CAMIÓN")
        super().describir()
        print(f"Carga máxima: {self.carga_max_ton} toneladas")

    def calcular_tarifa(self, km):
        # Tarifa por km + recargo según toneladas máximas
        return (km * 1.20) + (self.carga_max_ton * 5)


# Flota polimórfica
flota = [
    Bus("ABC-123", "Hino", 40, "Norte-Sur"),
    Taxi("XYZ-987", "Chevrolet", 4, "Cristian Alvear"),
    Camion("CAM-555", "Volvo", 2, 8),
]

km_recorridos = 20
print(f"=== Tarifario para un recorrido de {km_recorridos} km ===\n")
for v in flota:
    v.describir()
    print(f"Tarifa estimada: ${v.calcular_tarifa(km_recorridos):.2f}")
    print("-" * 40)

# Este programa demuestra:
# - REUTILIZACIÓN: la clase Vehiculo concentra lo común (placa, marca, capacidad).
# - ESPECIALIZACIÓN: cada hijo agrega atributos propios (ruta, conductor, carga).
# - POLIMORFISMO: el bucle llama describir() y calcular_tarifa() sin saber
#   qué tipo concreto de vehículo está procesando: cada objeto responde a su manera.
```
