## Ejercicio 1. Herencia básica

Vamos a crear una clase base llamada `Animal` y una clase hija llamada `Perro`.

### Instrucciones
1. Cree una clase `Animal`.
2. Agregue un atributo `nombre`.
3. Agregue un método `comer()`.
4. Cree una clase `Perro` que herede de `Animal`.
5. Agregue un método `ladrar()`.

### Preguntas guía
- ¿Qué elementos deben ir en la clase padre?
- ¿Qué elemento será exclusivo de la clase hija?

```python
# Ejercicio 1

# Clase Animal (clase padre)
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        print(f"{self.nombre} está comiendo.")


# Clase Perro (clase hija) - hereda de Animal
class Perro(Animal):
    def ladrar(self):
        print(f"{self.nombre} dice: ¡Guau, guau!")
```

```python
# Prueba del ejercicio 1
# Creamos un objeto Perro y probamos métodos heredados (comer)
# y métodos propios (ladrar)
mi_perro = Perro("Rocky")
mi_perro.comer()    # método heredado de Animal
mi_perro.ladrar()   # método propio de Perro
```

## Ejercicio 2. Herencia con atributos compartidos

Ahora trabajaremos con una clase `Persona` y una clase `Estudiante`.

### Instrucciones
1. Cree la clase `Persona` con los atributos:
   - nombre
   - edad
2. Agregue un método `mostrar_datos()`.
3. Cree la clase `Estudiante` que herede de `Persona`.
4. Agregue el atributo `carrera`.

### Preguntas guía
- ¿Qué atributos deben ser heredados?
- ¿Qué atributo solo pertenece a `Estudiante`?

```python
# Ejercicio 2

# Clase Persona
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def mostrar_datos(self):
        print(f"Nombre: {self.nombre} | Edad: {self.edad}")


# Clase Estudiante (hereda de Persona)
class Estudiante(Persona):
    def __init__(self, nombre, edad, carrera):
        # En este ejercicio aún no usamos super(): repetimos las
        # asignaciones del padre para verlo "a mano"
        self.nombre = nombre
        self.edad = edad
        self.carrera = carrera

    def mostrar_datos(self):
        print(f"Nombre: {self.nombre} | Edad: {self.edad} | Carrera: {self.carrera}")
```

```python
# Prueba del ejercicio 2
est = Estudiante("Cristian Alvear", 21, "Ingeniería en Software")
est.mostrar_datos()
```

## Ejercicio 3. Uso de `super()`

Vamos a mejorar el ejercicio anterior usando `super()`.

### Instrucciones
1. En la clase `Estudiante`, use `super().__init__(...)` para reutilizar el constructor de `Persona`.
2. Agregue un método para mostrar la carrera.

### Preguntas guía
- ¿Para qué sirve `super()`?
- ¿Qué ventaja tiene frente a volver a escribir los atributos heredados?

```python
# Ejercicio 3 - Mismo caso, ahora usando super()

class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def mostrar_datos(self):
        print(f"Nombre: {self.nombre} | Edad: {self.edad}")


class Estudiante(Persona):
    def __init__(self, nombre, edad, carrera):
        # super() llama al constructor de Persona y reutiliza
        # las asignaciones de nombre y edad
        super().__init__(nombre, edad)
        self.carrera = carrera

    def mostrar_carrera(self):
        print(f"{self.nombre} estudia {self.carrera}")
```

```python
# Prueba del ejercicio 3
est = Estudiante("María López", 19, "Medicina")
est.mostrar_datos()      # heredado
est.mostrar_carrera()    # propio
```

## Ejercicio 4. Polimorfismo básico

Ahora construiremos varias clases con un mismo método, pero con comportamientos diferentes.

### Instrucciones
1. Cree una clase base `Animal`.
2. Agregue un método `hacer_sonido()`.
3. Cree las clases `Perro` y `Gato`.
4. Redefina el método `hacer_sonido()` en cada clase.

### Preguntas guía
- ¿Qué significa redefinir un método?
- ¿Por qué este ejercicio representa polimorfismo?

```python
# Ejercicio 4 - Polimorfismo básico

class Animal:
    def hacer_sonido(self):
        print("El animal hace un sonido genérico.")


class Perro(Animal):
    def hacer_sonido(self):
        print("¡Guau, guau!")


class Gato(Animal):
    def hacer_sonido(self):
        print("¡Miau!")
```

```python
# Prueba del ejercicio 4
p = Perro()
g = Gato()
p.hacer_sonido()
g.hacer_sonido()
```

## Ejercicio 5. Polimorfismo con listas

Vamos a recorrer varios objetos en una misma lista.

### Instrucciones
1. Cree una lista con objetos de distintas clases.
2. Recorra la lista con un `for`.
3. Llame al mismo método en todos los objetos.

### Preguntas guía
- ¿Cómo responde cada objeto al mismo método?
- ¿Qué ventaja tiene esto en programas grandes?

```python
# Ejercicio 5 - Polimorfismo con listas

# Reutilizamos las clases del Ejercicio 4
animales = [Perro(), Gato(), Animal(), Perro(), Gato()]

# Llamamos al MISMO método en todos los objetos: cada uno responde
# a su manera. Eso es polimorfismo.
for a in animales:
    a.hacer_sonido()
```

## Ejercicio 6. Caso aplicado: Vehículos

Vamos a construir una jerarquía con una clase base `Vehiculo` y dos clases hijas.

### Instrucciones
1. Cree la clase `Vehiculo` con:
   - marca
   - modelo
   - método `mostrar_info()`
2. Cree la clase `Auto`.
3. Cree la clase `Moto`.
4. Use `super()` en ambas clases hijas.

### Preguntas guía
- ¿Qué atributos son comunes?
- ¿Qué atributos podrían ser específicos de cada clase?

```python
# Ejercicio 6 - Definición de clases

class Vehiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def mostrar_info(self):
        print(f"Marca: {self.marca} | Modelo: {self.modelo}")


class Auto(Vehiculo):
    def __init__(self, marca, modelo, num_puertas):
        super().__init__(marca, modelo)
        self.num_puertas = num_puertas

    def mostrar_info(self):
        super().mostrar_info()
        print(f"Puertas: {self.num_puertas}")


class Moto(Vehiculo):
    def __init__(self, marca, modelo, cilindrada):
        super().__init__(marca, modelo)
        self.cilindrada = cilindrada

    def mostrar_info(self):
        super().mostrar_info()
        print(f"Cilindrada: {self.cilindrada} cc")
```

```python
# Prueba del ejercicio 6
auto1 = Auto("Toyota", "Corolla", 4)
moto1 = Moto("Yamaha", "MT-03", 321)

auto1.mostrar_info()
print("-" * 30)
moto1.mostrar_info()
```

## Ejercicio 7. Sobrescritura de métodos

Vamos a redefinir un método heredado para que cada clase tenga un comportamiento particular.

### Instrucciones
1. En `Auto`, cree o redefina un método `encender()`.
2. En `Moto`, cree o redefina también el método `encender()`.
3. Compare los resultados.

### Preguntas guía
- ¿Qué diferencia hay entre heredar un método y sobrescribirlo?
- ¿En qué casos conviene sobrescribir?

```python
# Ejercicio 7 - Sobrescritura del método encender()
# Extendemos las clases Auto y Moto del ejercicio 6 añadiendo encender()

class AutoConEncendido(Auto):
    def encender(self):
        print(f"El auto {self.marca} {self.modelo} arranca girando la llave. Motor de 4 tiempos en marcha.")


class MotoConEncendido(Moto):
    def encender(self):
        print(f"La moto {self.marca} {self.modelo} arranca con el botón de encendido / patada.")
```

```python
# Prueba del ejercicio 7
auto = AutoConEncendido("Toyota", "Corolla", 4)
moto = MotoConEncendido("Yamaha", "MT-03", 321)

auto.encender()
moto.encender()

# Aunque ambos tienen un método llamado encender(), cada clase
# define su propio comportamiento: eso es sobrescritura.
```

## Ejercicio 8. `super()` dentro de un método

No solo se puede usar `super()` en el constructor. También puede utilizarse dentro de otros métodos.

### Instrucciones
1. Cree una clase `Empleado` con un método `presentarse()`.
2. Cree una clase `Docente` que herede de `Empleado`.
3. En `Docente`, redefina `presentarse()`.
4. Use `super().presentarse()` antes de agregar información nueva.

### Preguntas guía
- ¿Qué parte del método proviene de la clase padre?
- ¿Qué parte se agrega en la clase hija?

```python
# Ejercicio 8 - super() dentro de un método (no sólo en el constructor)

class Empleado:
    def __init__(self, nombre):
        self.nombre = nombre

    def presentarse(self):
        print(f"Hola, soy {self.nombre} y trabajo en esta institución.")


class Docente(Empleado):
    def __init__(self, nombre, materia):
        super().__init__(nombre)
        self.materia = materia

    def presentarse(self):
        # Primero ejecutamos el comportamiento del padre
        super().presentarse()
        # Y luego añadimos lo específico del docente
        print(f"Soy docente y dicto la materia de {self.materia}.")
```

```python
# Prueba del ejercicio 8
docente = Docente("Ana Castro", "Programación II")
docente.presentarse()
```

## Ejercicio 9. Herencia multinivel

Ahora trabajaremos con tres niveles de herencia.

### Instrucciones
1. Cree la clase `Persona`.
2. Cree la clase `Estudiante` que herede de `Persona`.
3. Cree la clase `EstudianteBecado` que herede de `Estudiante`.
4. Agregue atributos y un método para mostrar la información.

### Preguntas guía
- ¿Qué atributos vienen desde la clase base?
- ¿Qué atributos se agregan en cada nivel?

```python
# Ejercicio 9 - Herencia multinivel

class Persona:
    def __init__(self, nombre, cedula):
        self.nombre = nombre
        self.cedula = cedula

    def mostrar(self):
        print(f"Nombre: {self.nombre} | Cédula: {self.cedula}")


class Estudiante(Persona):
    def __init__(self, nombre, cedula, carrera):
        super().__init__(nombre, cedula)
        self.carrera = carrera

    def mostrar(self):
        super().mostrar()
        print(f"Carrera: {self.carrera}")


class EstudianteBecado(Estudiante):
    def __init__(self, nombre, cedula, carrera, tipo_beca, descuento):
        super().__init__(nombre, cedula, carrera)
        self.tipo_beca = tipo_beca
        self.descuento = descuento

    def mostrar(self):
        super().mostrar()
        print(f"Beca: {self.tipo_beca} ({self.descuento}% de descuento)")
```

```python
# Prueba del ejercicio 9
becado = EstudianteBecado(
    "Cristian Alvear", "0987654321", "Software",
    "Académica", 50
)
becado.mostrar()
```

## Ejercicio 10. Reto final 

Desarrolle un sistema con:

- una clase base `Empleado`
- una clase hija `Docente`
- una clase hija `Administrativo`

### Requisitos
- usar herencia,
- usar `super()`,
- sobrescribir al menos un método,
- crear varios objetos,
- recorrerlos en una lista para demostrar polimorfismo.

### Sugerencia
Puede usar atributos como:
- nombre
- sueldo
- departamento
- materia

```python
# Ejercicio 10 - Reto final
# Sistema con Empleado (base), Docente y Administrativo,
# usando herencia, super(), sobrescritura de métodos y polimorfismo.

class Empleado:
    def __init__(self, nombre, sueldo):
        self.nombre = nombre
        self.sueldo = sueldo

    def calcular_pago(self):
        return self.sueldo

    def mostrar_info(self):
        print(f"Nombre: {self.nombre}")
        print(f"Sueldo base: ${self.sueldo:.2f}")
        print(f"Pago final: ${self.calcular_pago():.2f}")


class Docente(Empleado):
    def __init__(self, nombre, sueldo, materia):
        super().__init__(nombre, sueldo)
        self.materia = materia

    def calcular_pago(self):
        # Los docentes reciben un 10% adicional
        return self.sueldo * 1.10

    def mostrar_info(self):
        print(">> DOCENTE")
        super().mostrar_info()
        print(f"Materia: {self.materia}")


class Administrativo(Empleado):
    def __init__(self, nombre, sueldo, departamento):
        super().__init__(nombre, sueldo)
        self.departamento = departamento

    def calcular_pago(self):
        # Los administrativos reciben un 5% adicional
        return self.sueldo * 1.05

    def mostrar_info(self):
        print(">> ADMINISTRATIVO")
        super().mostrar_info()
        print(f"Departamento: {self.departamento}")


# Lista polimórfica de empleados
empleados = [
    Docente("Ana Castro", 1500, "Programación II"),
    Docente("Luis Pérez", 1400, "Bases de Datos"),
    Administrativo("Carla Ríos", 1100, "Talento Humano"),
    Administrativo("Mario Soto", 1200, "Contabilidad"),
]

# Recorremos la lista llamando al MISMO método en cada objeto
for e in empleados:
    e.mostrar_info()
    print("-" * 40)
```
